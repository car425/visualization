```python
# Have separate files that list, per year, number of cats and dogs per FSA.  
# Concatenate these into one large dataframe.

xls_files = 'xls_files/by*.xls'
all_xls_files = glob(xls_files)
print(all_xls_files)
```

```python
xls_files_pets_df = []
rows_to_drop = [0, 1, 2, 3]

for file in all_xls_files:
    dataframe = pd.read_excel(file, header=None, index_col=None)
    dataframe = dataframe.dropna(axis = 1, how = 'all')
    dataframe = dataframe.drop(rows_to_drop, axis = 0)
    filename = os.path.basename(file) 
    name_parts = filename.split('-')
    year_and_xls = name_parts[-1]
    year_parts = year_and_xls.split('.')
    year = year_parts[-2]
    dataframe['Year'] = year
    xls_files_pets_df.append(dataframe)

Toronto_pets_df_concat = pd.concat(xls_files_pets_df, ignore_index=True)
Toronto_pets_df_concat.head(30)


```

```python
Toronto_pets_df_concat.columns = ['FSA', 'CAT', 'DOG', 'Total', 'Year']
```
